```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
one_hot = OneHotEncoder(handle_unknown="ignore")
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", one_hot, categorical_cols)
    ],
    remainder="passthrough"
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
```

```python
X_test_processed = preprocessor.transform(X_test)
```

```python
print("Before preprocessing:", X_train.shape)
print("After preprocessing :", X_train_processed.shape)
print("Test after preprocessing:", X_test_processed.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()
print("Number of processed features:", len(feature_names))
```

```python
feature_names
```

```python
for col in categorical_cols:
    print(col, "→", X_train[col].nunique(), "categories")
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numeric_features = [
    "count_floors_pre_eq",
    "age",
    "area_percentage",
    "height_percentage",
    "count_families"
]
```

```python
preprocessor_v2 = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", StandardScaler(), numeric_features)
    ],
    remainder="passthrough"
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
model = Pipeline([
    ("preprocessor", preprocessor_v2),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[1, 2, 3]
)

disp.plot()
plt.title("Confusion Matrix — Logistic Regression")
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor_v2),
    ("classifier", RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
param_grid = {
    "classifier__n_estimators": [100, 200],
    "classifier__min_samples_leaf": [1, 2, 5],
    "classifier__max_features": ["sqrt", "log2"]
}
```

```python
grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1,
    verbose=2
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("Best CV Macro F1:")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

y_test_pred = best_model.predict(X_test)
```

```python
final_accuracy = accuracy_score(y_test, y_test_pred)

print("Final Test Accuracy:", final_accuracy)
```

```python
print(classification_report(y_test, y_test_pred))
```

```python
final_cm = confusion_matrix(y_test, y_test_pred)

print(final_cm)
```

```python
final_disp = ConfusionMatrixDisplay(
    confusion_matrix=final_cm,
    display_labels=[1, 2, 3]
)

final_disp.plot()
plt.title("Final Random Forest - Confusion Matrix")
plt.show()
```
